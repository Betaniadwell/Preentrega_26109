```python
import pandas as pd
import gdown

# 1. FUNCIÓN DE ANÁLISIS
def quick_analysis(df):
    print("--- 1. DIMENSIONES ---")
    print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}")

    # Agregado: Lista de títulos de columnas
    print(f"Títulos de columnas: {df.columns.tolist()}")

    print("\n--- 2. DIAGNÓSTICO DE COLUMNAS ---")
    info_table = pd.DataFrame({
        'Tipo': df.dtypes,
        'Nulos': df.isnull().sum(),
        '% Nulos': (df.isnull().sum() / len(df) * 100).round(2),
        'Valores Únicos': df.nunique()
    })
    print(info_table)

    print("\n--- 3. INTEGRIDAD ---")
    print(f"Filas duplicadas totales: {df.duplicated().sum()}")

    print("\n--- 4. MUESTRA (SAMPLE) ---")
    display(df.sample(min(5, len(df))))

 #2. CONFIGURACIÓN DE DESCARGA
file_id = "1dpvV5iOAKyD7QZX8gwAZVZIefWGpnTKM"
url_download = f'https://drive.google.com/uc?id={file_id}'
output = 'ventas_descargado.csv'

try:
    #Descarga
    gdown.download(url_download, output, quiet=False)
    df_ventas = pd.read_csv(output)

    # ---  LÍNEAS DE LIMPIEZA ---
    # Limpiamos 'producto'
    df_ventas['producto'] = df_ventas['producto'].astype(str).str.strip().str.lower()

    df_ventas['precio'] = (df_ventas['precio']
                           .astype(str)
                           .str.replace('$', '', regex=False)
                           .str.replace(',', '', regex=False)
                           .str.strip())
    # Ejecutar el análisis con los datos ya limpios
    quick_analysis(df_ventas)

except Exception as e:
    print(f"Error: {e}")
```

## 2.2- Transformación de Datos: Productos de Alto Rendimiento

```python
# Observamos porcentaje de ventas
print(df_ventas[['precio', 'cantidad']].describe())
```

```python
# TABLA UNIFICA POR PRODUCTO ORDENADO DESDE MAYOR VENTAS

# 1. Agrupamos directamente desde la tabla de origen (df_ventas)
mejores_ventas = df_ventas.groupby('producto').agg(
    cantidad=('cantidad', 'sum'),
    total_ventas=('precio', 'sum')
).reset_index()

# 2. Creamos la columna 'prod_orden'
mejores_ventas['prod_reorden'] = mejores_ventas['producto']

# 3. Traemos la categoría (mapeo seguro)
if 'categoria' in df_ventas.columns:
    mapping_cat = df_ventas.groupby('producto')['categoria'].first()
    mejores_ventas['categoria'] = mejores_ventas['producto'].map(mapping_cat)
else:
    # Si no existe, creamos la columna vacía para que el paso final no falle
    mejores_ventas['categoria'] = "Sin Categoría"

# 4. Ordenamos por ventas
mejores_ventas = mejores_ventas.sort_values(by='total_ventas', ascending=False)

# 5. Selección final de columnas con .copy() para evitar avisos de memoria
columnas = ['categoria', 'cantidad', 'prod_reorden', 'total_ventas']
mejores_ventas = mejores_ventas[columnas].copy()

columnas = ['categoria', 'cantidad', 'prod_reorden', 'total_ventas']
mejores_ventas = mejores_ventas[columnas].copy().reset_index(drop=True)

mejores_ventas.index = mejores_ventas.index + 1


# 6. Verificamos
mejores_ventas.head(30)
```

```python
mejores_ventas.describe()
```

OBTENEMOS LOS CUARTILES SEGUN EL TOTAL DE VENTAS POR PRODUCTO

```python
# 1. Calculamos los límites (fíjate que quité el espacio en el nombre)
q1 = mejores_ventas['total_ventas'].quantile(0.25)
q2 = mejores_ventas['total_ventas'].quantile(0.50)
q3 = mejores_ventas['total_ventas'].quantile(0.75)

# 2. Definimos la función
def segmentar_por_cuartil(valor):
    if valor >= q3:
        return 'A - Oro (Top 25%)'
    elif valor >= q2:
        return 'B - Plata (50-75%)'
    elif valor >= q1:
        return 'C - Bronce (25-50%)'
    else:
        return 'D - Carbón (Bajo 25%)'

# 3. Aplicamos la segmentación directamente sobre mejores_ventas
mejores_ventas['rango'] = mejores_ventas['total_ventas'].apply(segmentar_por_cuartil)

# 4. Ordenamos por rango (A a D) y luego por ventas
resumen_cuartiles = mejores_ventas.sort_values(by=['rango', 'total_ventas'], ascending=[True, False])

# 5. Ajustar el índice para que empiece en 1
resumen_cuartiles.index = range(1, len(resumen_cuartiles) + 1)

display(resumen_cuartiles)
```

## 2.3- Agregación: Análisis de Ingresos por categoria



#### Ventas por Categoria          

```python
# 1. Agrupamiento (esto está perfecto)
df_resumen = df_ventas.groupby(["categoria"], as_index=False).agg(
    ventas_totales=("precio", "sum"),
    cant_ventas=("precio", "count"),
    avg_ventas=("precio", "mean")
)

# 2. ORDENAR (Guardamos el resultado ordenado directamente en resumen_categoria)
resumen_categoria = df_resumen.sort_values(by='ventas_totales', ascending=False)

# 3. REDONDEAR (Trabajamos sobre la tabla que YA ESTÁ ORDENADA)
# No uses df_resumen.copy() aquí, porque arruinas el orden anterior
resumen_categoria['avg_ventas'] = resumen_categoria['avg_ventas'].round(2)

# Resetear el índice para que el 1 sea el que más vendió
resumen_categoria.index = range(1, len(resumen_categoria) + 1)

# Visualizar
display(resumen_categoria)
```

Categorias por segmento

```python
# 1. Agrupamos por Categoría y Rango para contar los productos
resumen_cat_rango = mejores_ventas.groupby(['categoria', 'rango']).size().unstack(fill_value=0)

# 2. Agregamos un total por categoría para ver cuáles son las más pesadas
resumen_cat_rango['Total_Productos'] = resumen_cat_rango.sum(axis=1)

# 3. Ordenamos para que las categorías con más productos "Oro" salgan arriba
resumen_cat_rango = resumen_cat_rango.sort_values(by='A - Oro (Top 25%)', ascending=False)

display(resumen_cat_rango)
```

```python
# Procesos preliminares demostrativos para el punto N° 1

#COSTO TOTAL DE MARKETING POR PRODUCTO UNIFICADO

# MANIPULAMOS EL VALOR DE COSTO  CON EL FIN DE VISUALIZAR GRAFICOS MAS DEMOSTRATIVOS,
# CONSIDERANDO QUE EL COSTO ES POR CADA UNO DE LOS PRODUCTOS CORRESPONDIENTES
# MULTIPLICADOS ADEMAS POR UN NUMERO PARA LOGRAR RESULTADOS SIGNIFICATIVOS

df_marketing['costo'] = (df_marketing['costo'] * df_ventas['cantidad'] * 50 ).round(2)


# 1. Agrupar y resetear índice para alinear columnas
resultado = df_marketing.groupby('producto')['costo'].sum().sort_values(ascending=False).reset_index()

# 2. Hacer que el índice comience en 1
resultado.index = resultado.index + 1

# 3. Redondear valores
resultado = resultado.round(2)

# Estadísticas

clases_productos = df_marketing['producto'].nunique()
total_general = df_marketing['costo'].sum().round(2)

print(f"Total de clases de productos: {clases_productos}")
print(f"Costo total general: {total_general}")
print("\nDesglose por producto:")

# Mostrar tabla
resultado
```

 #####  RESOLUCION 1

```python
# OBTENEMOS CUANTO SE GASTO EN MARKETING POR CADA PRODUCTO
# MOSTRANDO ADEMAS EL TOTAL DE VENTAS POR PRODUCTO
# REPETIMOS UN REORDENAMIENTO GENERAL EN LAS TABLAS INVOLUCRADAS
# COMO MEDIDA PREVENTIVA

# 1. LIMPIEZA DE VENTAS
df_ventas['producto'] = df_ventas['producto'].astype(str).str.strip().str.lower() # Todo a minúsculas
df_ventas['precio'] = df_ventas['precio'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.strip()
df_ventas['precio'] = pd.to_numeric(df_ventas['precio'], errors='coerce')

ventas_resumen = df_ventas.groupby('producto').agg(
    cantidad=('cantidad', 'sum'),
    precio_total=('precio', 'sum'),
    categoria=('categoria', 'first')).reset_index()

# 2. LIMPIEZA DE MARKETING
df_marketing['producto'] = df_marketing['producto'].astype(str).str.strip().str.lower() # Todo a minúsculas
df_marketing['costo'] = pd.to_numeric(df_marketing['costo'], errors='coerce')

marketing_resumen = df_marketing.groupby('producto')['costo'].sum().reset_index()
marketing_resumen.columns = ['producto', 'marketing']

# 3. UNIÓN FINAL (Merge)
agrup_prod_vtas_mkt = pd.merge(marketing_resumen, ventas_resumen, on='producto', how='outer').fillna(0)
agrup_prod_vtas_mkt = agrup_prod_vtas_mkt.rename(columns={'precio_total': 'vtas_productos'})

# 4. FORMATEO
agrup_prod_vtas_mkt ['producto'] = agrup_prod_vtas_mkt['producto'].str.capitalize()

agrup_prod_vtas_mkt  = agrup_prod_vtas_mkt [['producto', 'categoria', 'cantidad', 'marketing', 'vtas_productos']]
agrup_prod_vtas_mkt  = agrup_prod_vtas_mkt.sort_values(by='vtas_productos', ascending=False).round(2).reset_index(drop=True)

agrup_prod_vtas_mkt.index = agrup_prod_vtas_mkt.index + 1

# RESULTADOS
print(f"Total clases productos: {agrup_prod_vtas_mkt['producto'].nunique()}")
print(f"Inversión Total Marketing: {agrup_prod_vtas_mkt['marketing'].sum().round(2)}")
agrup_prod_vtas_mkt
```

```python
total_productos = agrup_prod_vtas_mkt['cantidad'].sum()
total_ventas= agrup_prod_vtas_mkt['vtas_productos'].sum()
total_marketing= agrup_prod_vtas_mkt['marketing'].sum()
incidencia_mkt = (total_marketing / total_ventas) * 100
print(f"La cantidad total de productos  es: {total_productos}")
print(f"La cantidad total de ventas es: {total_ventas}")
print(f"La cantidad total de costo de marketing es: {total_marketing}")
print(f"La incidencia de marketing sobre el total de las ventas es: {incidencia_mkt.round(2)}%")
```

```python
# Agragamos columna 'ROAS',cuanto se recupero por cada peso invertido y columna
# inversion_sobre_vtas, informando que porcentaje sobre el total de las ventas se invirtio en marketing

# 1. Agrupamos y creamos la tabla base
analisis_vtas_mkt = agrup_prod_vtas_mkt.groupby('categoria').agg({
    'cantidad': 'sum',
    'marketing': 'sum',
    'vtas_productos': 'sum'
}).rename(columns={'vtas_productos': 'total_ventas'}).reset_index()

# 2. CALCULAMOS
analisis_vtas_mkt['roas'] = analisis_vtas_mkt['total_ventas'] / analisis_vtas_mkt['marketing']
analisis_vtas_mkt['inversion_sobre_vtas'] = (analisis_vtas_mkt['marketing'] / analisis_vtas_mkt['total_ventas']) * 100

# 3. ORDENAMOS (¡Cuidado aquí! No uses la variable vieja)
analisis_vtas_mkt= analisis_vtas_mkt.sort_values(by='total_ventas', ascending=False).reset_index(drop=True)
analisis_vtas_mkt.index = analisis_vtas_mkt.index + 1

# 4. FORMATO FINAL (Redondeos y símbolos)
analisis_vtas_mkt['total_ventas'] = analisis_vtas_mkt['total_ventas'].round(2)
analisis_vtas_mkt['marketing'] = analisis_vtas_mkt['marketing'].round(2)
analisis_vtas_mkt['roas'] = analisis_vtas_mkt['roas'].round(2)

# Formateamos el porcentaje como texto (esto debe ser lo último porque deja de ser número)
analisis_vtas_mkt['inversion_sobre_vtas'] = analisis_vtas_mkt['inversion_sobre_vtas'].map('{:.4f}%'.format)

print("✅ RESUMEN ACTUALIZADO")
display(analisis_vtas_mkt)
```

```python
import numpy as np

# 1. Creamos las columnas de cálculo primero
df_resultado = agrup_prod_vtas_mkt.copy()

# Calculamos ROAS (manejando divisiones por cero con np.where)
df_resultado['roas'] = np.where(
    df_resultado['marketing'] > 0,
    (df_resultado['vtas_productos'] / df_resultado['marketing']).round(2),
    0
)

# Calculamos Incidencia (Inversión sobre ventas)
df_resultado['inver_sobre_vtas'] = np.where(
    df_resultado['vtas_productos'] > 0,
    ((df_resultado['marketing'] / df_resultado['vtas_productos']) * 100).round(4),
    0
)

# 2. Ahora sí reordenamos y seleccionamos (todas las columnas ya existen)
columnas_ordenadas = ['producto', 'categoria', 'vtas_productos', 'marketing', 'roas', 'inver_sobre_vtas']
df_resultado = df_resultado[columnas_ordenadas]

# 3. Ordenamos por mayores ventas
df_resultado = df_resultado.sort_values(by='vtas_productos', ascending=False).reset_index(drop=True)
vtas_mkt_comparables = df_resultado.copy()

# 4. Ajustamos índice para la presentación
vtas_mkt_comparables.index = vtas_mkt_comparables.index + 1

# Mostrar resultado
print("Análisis de ROAS e Incidencia por Producto")
vtas_mkt_comparables
```

```python
%%writefile app.py
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import streamlit as st

# 1. Configuración del diseño de la página
st.set_page_config(page_title="Informe Estadístico Comercial", layout="wide")
st.title("📊 Panel de Control e Informe de Rendimiento Comercial")
st.markdown("---")

# 2. Carga de los DataFrames reales
@st.cache_data
def cargar_datos_csv():
    archivo_ventas = "datos_ventas_perfectos.csv"
    archivo_mkt = "datos_roi_perfectos.csv"

    if os.path.exists(archivo_ventas):
        df_q = pd.read_csv(archivo_ventas)
    else:
        import glob
        b = glob.glob("**/" + archivo_ventas, recursive=True)
        df_q = pd.read_csv(b) if b else pd.DataFrame()

    if os.path.exists(archivo_mkt):
        agrup_mkt = pd.read_csv(archivo_mkt)
    else:
        import glob
        b = glob.glob("**/" + archivo_mkt, recursive=True)
        agrup_mkt = pd.read_csv(b) if b else pd.DataFrame()

    return df_q, agrup_mkt

df_q, agrup_prod_vtas_mkt = cargar_datos_csv()

# 3. Creación de las Pestañas de Navegación Tradicionales
tab1, tab2, tab3 = st.tabs([
    "📊 Distribución y Análisis Estadístico",
    "🎯 Eficiencia de Marketing (ROI)",
    "🌌 Gráfico de Dispersión Comercial"
])

# --- PESTAÑA 1: DISTRIBUCIÓN Y BOXPLOT ---
with tab1:
    st.header("Análisis de Distribución de Ventas")

    if not df_q.empty:
        col_izq, col_der = st.columns(2)

        with col_izq:
            fig_hist, ax_hist = plt.subplots(figsize=(6, 5))
            sns.histplot(data=df_q, x='vtas_productos', bins="auto", ax=ax_hist, color="steelblue")
            ax_hist.set_title("Distribución de las Ventas (Histograma)")
            st.pyplot(fig_hist)

        with col_der:
            fig_box, ax_box = plt.subplots(figsize=(6, 5.5))
            sns.boxplot(data=df_q, y='vtas_productos', ax=ax_box, color="skyblue")
            ax_box.set_title("Distribución de las Ventas (Caja y Bigotes)")
            st.pyplot(fig_box)
    else:
        st.warning("⚠️ No se pudieron cargar los datos de ventas para esta pestaña.")

# --- PESTAÑA 2: EL GRÁFICO DE BARRAS DE ROI QUE YA QUEDÓ PERFECTO ---
with tab2:
    st.header("Eficiencia Real de Marketing")

    if not agrup_prod_vtas_mkt.empty:
        df_analisis = agrup_prod_vtas_mkt.copy()
        df_analisis["ROI_Marketing"] = df_analisis["vtas_productos"] / df_analisis["marketing"]
        df_analisis = df_analisis.sort_values(by="ROI_Marketing", ascending=False)

        fig_roi, ax_roi = plt.subplots(figsize=(10, 12))
        sns.set_theme(style="whitegrid")

        sns.barplot(data=df_analisis, x="ROI_Marketing", y="producto", palette="viridis", ax=ax_roi)

        plt.title("Eficiencia Real: Ventas generadas por cada Peso invertido en Marketing", fontsize=14, fontweight="bold")
        plt.xlabel("Multiplicador de Retorno (Ventas / Marketing)", fontsize=12)
        plt.ylabel("Producto", fontsize=12)
        plt.tight_layout()
        st.pyplot(fig_roi)
    else:
        st.warning("⚠️ No se encontraron datos para calcular el ROI.")

# --- PESTAÑA 3: GRÁFICO DE DISPERSIÓN COMERCIAL + CONCLUSIONES ---
with tab3:
    st.header("Análisis de Dispersión: Inversión vs. Ventas")
    st.markdown("Este gráfico permite visualizar qué productos tienen mayor tracción comercial en relación con su presupuesto asignado.")

    if not agrup_prod_vtas_mkt.empty:
        df_analisis = agrup_prod_vtas_mkt.copy()
        df_analisis["ROI_Marketing"] = df_analisis["vtas_productos"] / df_analisis["marketing"]

        if "marketing" in df_analisis.columns and "vtas_productos" in df_analisis.columns:
            fig_scat, ax_scat = plt.subplots(figsize=(10, 6))
            sns.set_theme(style="whitegrid")

            sns.scatterplot(
                data=df_analisis,
                x="marketing",
                y="vtas_productos",
                hue="producto",
                size="ROI_Marketing",
                sizes=(40, 400),
                palette="viridis",
                legend=False,
                ax=ax_scat
            )

            ax_scat.set_title("Relación entre Inversión en Publicidad y Ventas Totales", fontsize=14, fontweight="bold")
            ax_scat.set_xlabel("Presupuesto de Marketing ($)", fontsize=12)
            ax_scat.set_ylabel("Ventas de Productos ($)", fontsize=12)
            plt.tight_layout()
            st.pyplot(fig_scat)

            st.markdown("---")
            st.subheader("💡 Conclusiones y Diagnóstico Estratégico")

            col_info1, col_info2 = st.columns(2)
            with col_info1:
                st.success("""
                **🚀 Oportunidades de Crecimiento (Cuadrante Superior Izquierdo):**
                * Detectamos productos con **baja inversión en publicidad** que generan **ventas masivas** (cercanas a \$1.2M).
                * *Acción recomendada:* Incrementar el presupuesto de marketing en estos artículos de alta tracción orgánica.
                """)
                st.info("""
                **📈 Motores de Caja Estables (Línea Central):**
                * Las burbujas verdes demuestran un rendimiento constante entre \$600K y \$800K. Responden bien a aumentos de presupuesto, pero muestran un efecto de saturación al llegar al límite de su mercado.
                """)
            with col_info2:
                st.error("""
                **⚠️ Alertas de Pérdida / Ineficiencia (Zona Inferior):**
                * Las burbujas amarillas y pequeñas representan productos que, a pesar de recibir inversión constante (\$5K+), devuelven muy pocas ventas (menos de \$400K).
                * *Acción recomendada:* Pausar o reestructurar de inmediato las campañas publicitarias de estos artículos.
                """)
        else:
            st.info("💡 El archivo debe contener las columnas 'marketing' y 'vtas_productos'.")
```
