```python
# TABLA UNIFICA POR PRODUCTO ORDENADO DESDE MAYOR VENTAS

# 1. Agrupamos directamente desde la tabla de origen (df_ventas)
mejores_ventas = df_ventas.groupby('producto').agg(
    cantidad=('cantidad', 'sum'),
    total_ventas=('precio', 'sum')
).reset_index()

# 2. Creamos la columna 'prod_orden'
mejores_ventas['prod_reorden'] = mejores_ventas['producto']

# 3. Traemos la categoría (mapeo seguro)
if 'categoria' in df_ventas.columns:
    mapping_cat = df_ventas.groupby('producto')['categoria'].first()
    mejores_ventas['categoria'] = mejores_ventas['producto'].map(mapping_cat)
else:
    # Si no existe, creamos la columna vacía para que el paso final no falle
    mejores_ventas['categoria'] = "Sin Categoría"

# 4. Ordenamos por ventas
mejores_ventas = mejores_ventas.sort_values(by='total_ventas', ascending=False)

# 5. Selección final de columnas con .copy() para evitar avisos de memoria
columnas = ['categoria', 'cantidad', 'prod_reorden', 'total_ventas']
mejores_ventas = mejores_ventas[columnas].copy()

columnas = ['categoria', 'cantidad', 'prod_reorden', 'total_ventas']
mejores_ventas = mejores_ventas[columnas].copy().reset_index(drop=True)

mejores_ventas.index = mejores_ventas.index + 1


# 6. Verificamos
mejores_ventas.head(30)
```

```python
mejores_ventas.describe()
```

OBTENEMOS LOS CUARTILES SEGUN EL TOTAL DE VENTAS POR PRODUCTO

```python
# 1. Calculamos los límites (fíjate que quité el espacio en el nombre)
q1 = mejores_ventas['total_ventas'].quantile(0.25)
q2 = mejores_ventas['total_ventas'].quantile(0.50)
q3 = mejores_ventas['total_ventas'].quantile(0.75)

# 2. Definimos la función
def segmentar_por_cuartil(valor):
    if valor >= q3:
        return 'A - Oro (Top 25%)'
    elif valor >= q2:
        return 'B - Plata (50-75%)'
    elif valor >= q1:
        return 'C - Bronce (25-50%)'
    else:
        return 'D - Carbón (Bajo 25%)'

# 3. Aplicamos la segmentación directamente sobre mejores_ventas
mejores_ventas['rango'] = mejores_ventas['total_ventas'].apply(segmentar_por_cuartil)

# 4. Ordenamos por rango (A a D) y luego por ventas
resumen_cuartiles = mejores_ventas.sort_values(by=['rango', 'total_ventas'], ascending=[True, False])

# 5. Ajustar el índice para que empiece en 1
resumen_cuartiles.index = range(1, len(resumen_cuartiles) + 1)

display(resumen_cuartiles)
```

## 2.3- Agregación: Análisis de Ingresos por categoria



#### Ventas por Categoria          

```python
# 1. Agrupamiento
df_resumen = df_ventas.groupby(["categoria"], as_index=False).agg(
    ventas_totales=("precio", "sum"),
    cant_ventas=("precio", "count"),
    avg_ventas=("precio", "mean")
)

# 2. ORDENAR (Guardamos el resultado ordenado directamente en resumen_categoria)
resumen_categoria = df_resumen.sort_values(by='ventas_totales', ascending=False)

# 3. REDONDEAR (Trabajamos sobre la tabla que YA ESTÁ ORDENADA)

resumen_categoria['avg_ventas'] = resumen_categoria['avg_ventas'].round(2)

# Resetear el índice para que el 1 sea el que más vendió
resumen_categoria.index = range(1, len(resumen_categoria) + 1)

# Visualizar
display(resumen_categoria)
```

Categorias por segmento

```python
# 1. Agrupamos por Categoría y Rango para contar los productos
resumen_cat_rango = mejores_ventas.groupby(['categoria', 'rango']).size().unstack(fill_value=0)

# 2. Agregamos un total por categoría para ver cuáles son las más pesadas
resumen_cat_rango['Total_Productos'] = resumen_cat_rango.sum(axis=1)

# 3. Ordenamos para que las categorías con más productos "Oro" salgan arriba
resumen_cat_rango = resumen_cat_rango.sort_values(by='A - Oro (Top 25%)', ascending=False)

display(resumen_cat_rango)
```

```python
# Procesos preliminares demostrativos para el punto N° 1

#COSTO TOTAL DE MARKETING POR PRODUCTO UNIFICADO

# MANIPULAMOS EL VALOR DE COSTO  CON EL FIN DE VISUALIZAR GRAFICOS MAS DEMOSTRATIVOS,
# CONSIDERANDO QUE EL COSTO ES POR CADA UNO DE LOS PRODUCTOS CORRESPONDIENTES
# MULTIPLICADOS ADEMAS POR UN NUMERO PARA LOGRAR RESULTADOS SIGNIFICATIVOS

df_marketing['costo'] = (df_marketing['costo'] * df_ventas['cantidad'] * 50 ).round(2)


# 1. Agrupar y resetear índice para alinear columnas
resultado = df_marketing.groupby('producto')['costo'].sum().sort_values(ascending=False).reset_index()

# 2. Hacer que el índice comience en 1
resultado.index = resultado.index + 1

# 3. Redondear valores
resultado = resultado.round(2)

# Estadísticas

clases_productos = df_marketing['producto'].nunique()
total_general = df_marketing['costo'].sum().round(2)

print(f"Total de clases de productos: {clases_productos}")
print(f"Costo total general: {total_general}")
print("\nDesglose por producto:")

# Mostrar tabla
resultado
```

```python
# OBTENEMOS CUANTO SE GASTO EN MARKETING POR CADA PRODUCTO
# MOSTRANDO ADEMAS EL TOTAL DE VENTAS POR PRODUCTO
# REPETIMOS UN REORDENAMIENTO GENERAL EN LAS TABLAS INVOLUCRADAS

# 1. LIMPIEZA DE VENTAS
df_ventas['producto'] = df_ventas['producto'].astype(str).str.strip().str.lower() # Todo a minúsculas
df_ventas['precio'] = df_ventas['precio'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.strip()
df_ventas['precio'] = pd.to_numeric(df_ventas['precio'], errors='coerce')

ventas_resumen = df_ventas.groupby('producto').agg(
    cantidad=('cantidad', 'sum'),
    precio_total=('precio', 'sum'),
    categoria=('categoria', 'first')).reset_index()

# 2. LIMPIEZA DE MARKETING
df_marketing['producto'] = df_marketing['producto'].astype(str).str.strip().str.lower() # Todo a minúsculas
df_marketing['costo'] = pd.to_numeric(df_marketing['costo'], errors='coerce')

marketing_resumen = df_marketing.groupby('producto')['costo'].sum().reset_index()
marketing_resumen.columns = ['producto', 'marketing']

# 3. UNIÓN FINAL (Merge)
agrup_prod_vtas_mkt = pd.merge(marketing_resumen, ventas_resumen, on='producto', how='outer').fillna(0)
agrup_prod_vtas_mkt = agrup_prod_vtas_mkt.rename(columns={'precio_total': 'vtas_productos'})

# 4. FORMATEO
agrup_prod_vtas_mkt ['producto'] = agrup_prod_vtas_mkt['producto'].str.capitalize()

agrup_prod_vtas_mkt  = agrup_prod_vtas_mkt [['producto', 'categoria', 'cantidad', 'marketing', 'vtas_productos']]
agrup_prod_vtas_mkt  = agrup_prod_vtas_mkt.sort_values(by='vtas_productos', ascending=False).round(2).reset_index(drop=True)

agrup_prod_vtas_mkt.index = agrup_prod_vtas_mkt.index + 1

# RESULTADOS
print(f"Total clases productos: {agrup_prod_vtas_mkt['producto'].nunique()}")
print(f"Inversión Total Marketing: {agrup_prod_vtas_mkt['marketing'].sum().round(2)}")
agrup_prod_vtas_mkt
```

```python
%%writefile app.py
import os
import glob
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import streamlit as st
import plotly.graph_objects as go

# 1. Configuración del diseño de la página
st.set_page_config(page_title="Informe Estadístico Comercial", layout="wide")
st.title("📊 Panel de Control e Informe de Rendimiento Comercial")
st.markdown("---")

# 2. Carga de los DataFrames reales
def cargar_datos_csv():
    archivo_ventas = "datos_ventas_perfectos.csv"
    archivo_mkt = "datos_agrupados_marketing.csv"  # El nuevo archivo independiente

    def buscar_y_leer(nombre_archivo):
        if os.path.exists(nombre_archivo):
            return pd.read_csv(nombre_archivo)
        b = glob.glob("**/" + nombre_archivo, recursive=True)
        if b:
            return pd.read_csv(b[0])  # Toma la primera coincidencia encontrada en la lista
        return pd.DataFrame()

    df_q = buscar_y_leer(archivo_ventas)
    agrup_mkt = buscar_y_leer(archivo_mkt)
    return df_q, agrup_mkt

df_q, agrup_prod_vtas_mkt = cargar_datos_csv()

# 3. Creación de las 4 Pestañas de Navegación
tab1, tab2, tab3, tab4 = st.tabs([
    "📊 Distribución y Análisis Estadístico",
    "🎯 Detección de Atípicos (QRI)",
    "📈 Eficiencia de Marketing (ROI)",
    "🌌 Gráfico de Dispersión Comercial"
])

# --- PESTAÑA 1: DISTRIBUCIÓN Y BOXPLOT ---
with tab1:
    st.header("Análisis de Distribución de Ventas")

    if not df_q.empty:
        col_izq, col_der = st.columns(2)
        col_vta = 'vtas_productos' if 'vtas_productos' in df_q.columns else df_q.columns[0]

        with col_izq:
            fig_hist, ax_hist = plt.subplots(figsize=(6, 5))
            sns.histplot(data=df_q, x=col_vta, bins="auto", ax=ax_hist, color="steelblue")
            ax_hist.set_title("Distribución de las Ventas (Histograma)")
            st.pyplot(fig_hist)

        with col_der:
            fig_box, ax_box = plt.subplots(figsize=(6, 5.5))
            sns.boxplot(data=df_q, y=col_vta, ax=ax_box, color="skyblue")
            ax_box.set_title("Distribución de las Ventas (Caja y Bigotes)")
            st.pyplot(fig_box)
    else:
        st.warning("⚠️ No se pudieron cargar los datos de ventas para esta pestaña.")


# --- NUEVA PESTAÑA 2: DETECCIÓN DE ATÍPICOS (QRI) ---
with tab2:
    st.header("🎯 Detección de Atípicos (Rango Intercuartílico)")

    if not df_q.empty:
        col_vta = 'vtas_productos' if 'vtas_productos' in df_q.columns else df_q.columns[0]

        # 1. Cálculos estadísticos en tiempo real
        q1 = df_q[col_vta].quantile(0.25).round(2)
        q2 = df_q[col_vta].quantile(0.50).round(2)
        q3 = df_q[col_vta].quantile(0.75).round(2)
        v_min = round(df_q[col_vta].min(), 2)
        v_max = round(df_q[col_vta].max(), 2)

        iqr_p = q3 - q1
        QRI = q3 + (1.5 * iqr_p)

        # Filtrado de registros que superan el límite QRI
        atipicos = df_q[df_q[col_vta] > QRI].copy()

        # Diseño de dos columnas: Datos a la izquierda, Gráfico a la derecha
        col_datos, col_grafico = st.columns(2)

        with col_datos:
            st.subheader("📋 Resumen de Intervalos")
            st.markdown(f"""
            * **Mínimo:** ${v_min:,.2f}
            * **Primer Cuartil (q1):** ${q1:,.2f}
            * **Mediana (q2):** ${q2:,.2f}
            * **Tercer Cuartil (q3):** ${q3:,.2f}
            * **Máximo:** ${v_max:,.2f}
            """)
            st.info(f"**Línea Límite de Atípicos (QRI):** ${QRI:,.2f}")

            # Muestra cuántos registros superan el límite
            if not atipicos.empty:
                st.error(f"⚠️ Se detectaron **{len(atipicos)}** registros atípicos por encima del límite.")
            else:
                st.success("✅ No se detectaron valores atípicos superiores.")

        with col_grafico:
            # 2. Renderizado del Gráfico de Cajas personalizado
            fig_qri, ax_qri = plt.subplots(figsize=(10, 4.5))
            sns.set_theme(style="whitegrid")

            sns.boxplot(
                x=df_q[col_vta],
                color="skyblue",
                flierprops={"markerfacecolor": "red", "marker": "o", "markersize": 8},
                ax=ax_qri
            )

            # Línea roja de corte
            ax_qri.axvline(QRI, color='red', linestyle='--', linewidth=2, label=f'Límite QRI: {QRI:,.2f}')

            ax_qri.set_title('Distribución de Ventas con Límite de Atípicos', fontsize=14, pad=15)
            ax_qri.set_xlabel('Ventas de Productos ($)', fontsize=12)
            ax_qri.legend()

            st.pyplot(fig_qri)

        # --- NUEVA SECCIÓN: TABLA DE PRODUCTOS ATÍPICOS ---
        st.markdown("---")
        st.subheader("🔍 Listado Detallado de Ventas Atípicas")

        if not atipicos.empty:
            st.markdown("Los siguientes registros representan transacciones o productos con un volumen de ventas excepcionalmente alto:")

            # Ordenamos de mayor a menor venta para destacar los más importantes
            atipicos_ordenados = atipicos.sort_values(by=col_vta, ascending=False)

            # Mostramos la tabla formateada con Streamlit
            st.dataframe(
                atipicos_ordenados,
                use_container_width=True,
                column_config={
                    col_vta: st.column_config.NumberColumn(
                        "Ventas de Productos",
                        format="$%,.2f"
                    )
                }
            )
        else:
            st.info(f"No hay filas para mostrar ya que ningún registro supera el límite estadístico de {QRI:,.2f}.")

    else:
        st.warning("⚠️ No hay datos disponibles para el análisis QRI.")


# --- PESTAÑA 3: EFICIENCIA DE MARKETING (ROI) CON GRÁFICO NUEVO ---
with tab3:
    st.header("📈 Eficiencia de Marketing (ROI)")

    if not agrup_prod_vtas_mkt.empty:
        st.subheader("Análisis de Inversión: Ventas vs Marketing (Top 5 Productos)")

        # Preparación de datos basándose en tu DataFrame agrupado de Colab
        top_smart = agrup_prod_vtas_mkt.head(5).copy()
        top_smart['marketing_visual'] = top_smart['marketing'] * 50

        # Construcción del gráfico interactivo Plotly
        fig = go.Figure()

        # Barra de Ventas (Azul claro de fondo)
        fig.add_trace(go.Bar(
            x=top_smart['producto'],
            y=top_smart['vtas_productos'],
            name='Ventas Totales',
            marker_color='#3498db',
            opacity=0.3,
            text=top_smart['vtas_productos'],
            texttemplate='$%{text:,.2s}',
            textposition='outside',
            textfont=dict(color='#2a3f5f', size=12)
        ))

        # Barra de Marketing (Rosado - Impacto x50)
        fig.add_trace(go.Bar(
            x=top_smart['producto'],
            y=top_smart['marketing_visual'],
            name='Marketing (Impacto x50)',
            marker_color='#FF2266',
            text=top_smart['marketing_visual'],
            texttemplate='<b>$ %{text:,.0f}</b>',
            textposition='inside',
            insidetextanchor='middle',
            insidetextfont=dict(color='white', size=11),
            customdata=top_smart['marketing'],
            hovertemplate='<b>Gasto Real:</b> $%{customdata:,.2f}<extra></extra>'
        ))

        # Diseño Final del Layout para Web
        fig.update_layout(
            barmode='overlay',
            template="plotly_white",
            xaxis=dict(tickangle=0, title="Productos"),
            yaxis=dict(tickformat="$,.0s", title="Monto ($)"),
            legend=dict(orientation="h", y=1.1, x=0.5, xanchor="center"),
            margin=dict(l=40, r=40, t=40, b=40),
            height=500
        )

        # Ajuste de grosores
        fig.update_traces(width=0.6, selector=dict(name='Ventas Totales'))
        fig.update_traces(width=0.45, selector=dict(name='Marketing (Impacto x50)'))

        # Despliegue responsivo de Plotly en Streamlit
        st.plotly_chart(fig, use_container_width=True)

        # Despliegue de la nueva tabla agregada abajo del gráfico
        with st.expander("🔍 Ver Tabla de Datos de Marketing y Ventas (Subida desde Git)"):
            st.dataframe(agrup_prod_vtas_mkt, use_container_width=True)
    else:
        st.warning("⚠️ No se encontró el archivo 'datos_agrupados_marketing.csv'. Verifica haberlo subido desde Colab.")


# --- PESTAÑA 4: GRÁFICO DE DISPERSIÓN COMERCIAL ---
with tab4:
    st.header("🌌 Gráfico de Dispersión Comercial: Ventas vs Marketing")

    if not agrup_prod_vtas_mkt.empty:
        st.markdown("Este análisis permite evaluar de forma relacional qué impacto real genera la inversión en marketing sobre el volumen total de ventas por producto. Pasa el cursor sobre los puntos para inspeccionar cada registro.")

        # Filtramos el Top 3 para dejarles el texto fijo
        top_3_productos = agrup_prod_vtas_mkt.head(3)['producto'].tolist()

        # Creamos la lista de etiquetas fijas
        textos_visibles = [
            prod if prod in top_3_productos else ""
            for prod in agrup_prod_vtas_mkt['producto']
        ]

        fig_scatter = go.Figure()

        fig_scatter.add_trace(go.Scatter(
            x=agrup_prod_vtas_mkt['marketing'],
            y=agrup_prod_vtas_mkt['vtas_productos'],
            mode='markers+text',
            text=textos_visibles,
            textposition="top center",
            # FORZAMOS COLOR BLANCO DIRECTO PARA EL MODO OSCURO
            textfont=dict(
                size=12,
                color="white",  # Cambiado a blanco directo para legibilidad total
                family="Arial Black"
            ),
            marker=dict(
                size=14,
                color=agrup_prod_vtas_mkt['vtas_productos'],
                colorscale='Viridis',
                showscale=True,
                colorbar=dict(title="Ventas ($)", tickformat="$,.0s")
            ),
            hovertemplate="<b>📦 Producto:</b> %{customdata}<br>" +
                          "<b>💰 Inversión Mkt:</b> $%{x:,.2f}<br>" +
                          "<b>📈 Ventas Totales:</b> $%{y:,.2f}<extra></extra>",
            customdata=agrup_prod_vtas_mkt['producto']
        ))

        # Ajustamos el layout para que sea compatible con entornos oscuros
        fig_scatter.update_layout(
            template="plotly_dark",  # Cambiamos a plantilla oscura para que combine con tu entorno web
            xaxis=dict(
                title="Inversión en Marketing ($)",
                tickformat="$,.0f",
                gridcolor="rgba(255,255,255,0.1)"
            ),
            yaxis=dict(
                title="Ventas del Producto ($)",
                tickformat="$,.0f",
                gridcolor="rgba(255,255,255,0.1)"
            ),
            height=550,
            margin=dict(l=50, r=40, t=40, b=50)
        )

        st.plotly_chart(fig_scatter, use_container_width=True)
    else:
```
